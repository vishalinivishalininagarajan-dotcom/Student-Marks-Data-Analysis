```python
students = ["Abdul", "Harsitha", "Vishalini", "Sirii", "Sangee"]

marks = [85, 90, 78, 95, 88]

average = sum(marks) / len(marks)

print("Student Names:", students)
print("Student Marks:", marks)
print("Average Marks:", average)
print("Highest Marks:", max(marks))
print("Lowest Marks:", min(marks))
```

```python
for i in range(len(students)):
    print(students[i], "=", marks[i])
```

```python
for i in range(len(students)):
    if marks[i] >= 40:
        print(students[i], "- PASS")
    else:
        print(students[i], "- FAIL")
```

```python
import matplotlib.pyplot as plt

plt.bar(students, marks)

plt.xlabel("Student Names")
plt.ylabel("Marks")
plt.title("Student Marks Analysis")

plt.show()
```

```python
topper_index = marks.index(max(marks))

print("Class Topper:", students[topper_index])
print("Highest Mark:", max(marks))
```

```python
print(" STUDENT ANALYSIS REPORT ")

print("Total Students:", len(students))
print("Class Average:", average)
print("Highest Mark:", max(marks))
print("Lowest Mark:", min(marks))

print("Topper:", students[marks.index(max(marks))])

print(" ANALYSIS COMPLETED ")
```
